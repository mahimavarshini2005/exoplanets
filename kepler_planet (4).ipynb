```python
tpf = search_targetpixelfile('KIC 6922244', author="Kepler", cadence="long", quarter=4).download()
```

```python
tpf.plot()
```

```python
import matplotlib.pyplot as plt

plt.close('all')    # Close all existing figures
tpf.plot();         # Create one new plot
plt.show()
```

```python
lc = tpf.to_lightcurve(aperture_mask=tpf.pipeline_mask);
```

```python
import matplotlib.pyplot as plt

plt.close('all')    # Close all existing figures
lc.plot();          # Create one new plot
plt.show()
```

```python
flat_lc = lc.flatten(window_length=401) 
plt.close('all') 
flat_lc.plot();
import matplotlib.pyplot as plt
plt.show()
```

```python
folded_lc = flat_lc.fold(period=3.5225)
plt.close('all') 
folded_lc.plot();
import matplotlib.pyplot as plt
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.close('all')

binned_lc = folded_lc.bin(bins=10)
binned_lc.plot()

plt.show()
```

```python
import numpy as np
periodogram = flat_lc.to_periodogram(method="bls", period=np.arange(1,10,0.001))
```

```python
import matplotlib.pyplot as plt #note the period (d) with the largest peak

plt.close('all')
periodogram.plot()
plt.show()
```
